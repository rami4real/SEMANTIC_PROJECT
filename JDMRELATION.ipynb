```python
import requests
from urllib.parse import quote

url = "https://jdm-api.demo.lirmm.fr/v0/"

RELATION_ID_MAP = {
    "r_depict": 172,
    "r_has_causatif": 42,
    "r_has_prop": 153,
    "r_holo": 10,
    "r_lieu": 15,
    "r_object>mater": 50,
    "r_own-1": 122,
    "r_processus>agent-1": 137,
    "r_processus>instr": 80,
    "r_processus>patient-1": 138,
    "r_quantificateur-1": 174,
    "r_has_social_tie_with": 113,
    "r_has_topic": 142
}

def relationFromNoeudToNoeud(node1, relationId, node2):
    req = (
        url
        + "relations/from/"
        + quote(node1, safe="")
        + "/to/"
        + quote(node2, safe="")
        + "?types_ids="
        + str(relationId)
    )
    return requests.get(req).json()

def tester_relations_dict(A, B, rel_map):
    print(f"\nRelations entre '{A}' et '{B}':")

    for rel_name, relid in rel_map.items():
        data = relationFromNoeudToNoeud(A, relid, B)

        if not data.get("relations"):
            print(f"  {rel_name} : aucune relation")
        else:
            poids = data["relations"][0]["w"]
            print(f"  {rel_name} : poids = {poids}")
```
